```python
def generate_meal_combinations(template, main_dishes, side_dishes=None, beverages=None, desserts=None):
    # Handle optional lists
    side_dishes = side_dishes if side_dishes else [""]
    beverages = beverages if beverages else [""]
    desserts = desserts if desserts else [""]

    # Generate all possible combinations using nested list comprehensions
    meal_combinations = [
        template.format(main=main, side=side, beverage=beverage, dessert=dessert)
        for main in main_dishes
        for side in side_dishes
        for beverage in beverages
        for dessert in desserts
    ]

    return meal_combinations

# Example inputs
template = "{main} paired with {side}, {beverage}, and {dessert}"
main_dishes = ['Roasted duck', 'Mushroom risotto', 'Lobster tail']
side_dishes = ['Truffle fries', 'Quinoa salad', 'Garlic spinach']
beverages = ['Orange juice', 'Sparkling water', 'Artisanal soda']
desserts = ['Crème brûlée', 'Panna cotta', 'Macarons']

# Generate and print meal combinations
meals = generate_meal_combinations(template, main_dishes, side_dishes, beverages, desserts)
for meal in meals:
    print(meal)

```
