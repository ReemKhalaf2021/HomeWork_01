```python
def generate_meal_combinations(template, main_dishes, side_dishes=None, desserts=None):
    # Handle optional lists
    side_dishes = side_dishes if side_dishes else [""]
    desserts =  desserts if desserts else [""]

    # Generate all possible combinations using nested list comprehensions
    meal_combinations = [
        template.format(main=main, side=side, dessert=dessert)
        for main in main_dishes
        for side in side_dishes
        for dessert in desserts
    ]

    return meal_combinations

# Example inputs
template = "{main} accompanied by {side} and {dessert}"
main_dishes = ['Stuffed peppers', 'Tofu stir-fry']
side_dishes = ['Herb couscous', 'Roasted vegetables']
desserts = ['Lemon tart', 'Berry sorbet']

# Generate and print meal combinations
meals = generate_meal_combinations(template, main_dishes, side_dishes, desserts)
for meal in meals:
    print(meal)

```
